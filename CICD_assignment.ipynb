```python
//Create a Simple Python Application
def greet(name):
    return f"Hello, {name}!"

if __name__ == "__main__":
    print(greet("World"))

  //add a test file test_app.py
import unittest
from app import greet

class TestApp(unittest.TestCase):
    def test_greet(self):
        self.assertEqual(greet("Alice"), "Hello, Alice!")

if __name__ == "__main__":
    unittest.main()
//Set Up a GitHub Repository
git init
git remote add origin https://github.com/123pava/assignment.git
git add .
git commit -m "Initial commit"
git push -u origin main

//Configure GitHub Actions for CI
//Create a workflow file at .github/workflows/python-ci.yml:
name: Python CI

on:
  push:
    branches: [ main ]
  pull_request:
    branches: [ main ]

jobs:
  build:
    runs-on: ubuntu-latest

    steps:
    - name: Checkout code
      uses: actions/checkout@v3

    - name: Set up Python
      uses: actions/setup-python@v4
      with:
        python-version: '3.10'

    - name: Install dependencies
      run: |
        python -m pip install --upgrade pip
        pip install pytest

    - name: Run tests
      run: |
        pytest test_app.py
//push workflow
git add .github/workflows/python-ci.yml
git commit -m "Add CI workflow"
git push

```

```python
//Create a Simple REST API with Flask

from flask import Flask, jsonify, request
import sqlite3

app = Flask(__name__)

def init_db():
    conn = sqlite3.connect('data.db')
    conn.execute('CREATE TABLE IF NOT EXISTS items (id INTEGER PRIMARY KEY, name TEXT)')
    conn.commit()
    conn.close()

@app.route('/items', methods=['POST'])
def add_item():
    data = request.get_json()
    conn = sqlite3.connect('data.db')
    cursor = conn.cursor()
    cursor.execute('INSERT INTO items (name) VALUES (?)', (data['name'],))
    conn.commit()
    conn.close()
    return jsonify({'message': 'Item added'}), 201

@app.route('/items', methods=['GET'])
def get_items():
    conn = sqlite3.connect('data.db')
    cursor = conn.cursor()
    cursor.execute('SELECT * FROM items')
    items = cursor.fetchall()
    conn.close()
    return jsonify(items)

if __name__ == '__main__':
    init_db()
    app.run(debug=True)

    //run locally
python app.py

- Create a Postman collection with:
- POST /items → Add an item
- GET /items → Retrieve items
- Save the collection as api-tests.postman_collection.json

//cli testing

npm install -g newman
newman run api-tests.postman_collection.json
npm install -g newman
python app.py &  # Start Flask API in background
sleep 5          # Wait for server to start
newman run api-tests.postman_collection.json

```

7.Continuous Deployment (CD)

Task 7: Automate Deployment to Staging Environment

Objective: Automate the deployment of an application to a staging environment after a successful build and
test.

Tools: Jenkins, AWS CodeDeploy, Docker, Kubernetes

Instructions7
 Create a simple application (Node.js, Python, or Java)
 Set up a CI pipeline that builds the application, runs tests, and deploys it to a staging environment (using
AWS, Docker, or Kubernetes)
 Document the pipeline setup, and take screenshots of the deployment logs
 Ensure the application is live and accessible from the staging environment

```python
//Create a Simple Node.js Application
const express = require('express');
const app = express();
const port = process.env.PORT || 3000;

app.get('/', (req, res) => {
  res.send('Hello from the staging environment!');
});

app.listen(port, () => {
  console.log(`App running on port ${port}`);
});

//json file
{
  "name": "cd-staging-demo",
  "version": "1.0.0",
  "scripts": {
    "start": "node app.js",
    "test": "echo \"Running tests...\" && exit 0"
  },
  "dependencies": {
    "express": "^4.18.2"
  }
}

//docker file
FROM node:18

WORKDIR /usr/src/app

COPY package*.json ./
RUN npm install

COPY . .

EXPOSE 3000
CMD ["npm", "start"]

// pipeline configuration

pipeline {
    agent any

    tools {
        nodejs "NodeJS 18"
    }

    environment {
        IMAGE_NAME = "cd-staging-node"
        CONTAINER_NAME = "staging-node-app"
    }

    stages {
        stage('Checkout') {
            steps {
                git 'https://github.com/your-username/cd-staging-demo.git'
            }
        }

        stage('Install Dependencies') {
            steps {
                sh 'npm install'
            }
        }

        stage('Run Tests') {
            steps {
                sh 'npm test'
            }
        }

        stage('Build Docker Image') {
            steps {
                sh 'docker build -t $IMAGE_NAME .'
            }
        }

        stage('Deploy to Staging') {
            steps {
                sh '''
                docker stop $CONTAINER_NAME || true
                docker rm $CONTAINER_NAME || true
                docker run -d --name $CONTAINER_NAME -p 8080:3000 $IMAGE_NAME
                '''
            }
        }
    }

    post {
        success {
            echo 'Deployment to staging successful!'
        }
        failure {
            echo 'Build or deployment failed.'
        }
    }
}
```
